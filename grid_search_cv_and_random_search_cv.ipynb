```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
X=df.drop('species',axis=1)
y=df['species']
```

```python
from sklearn.svm import SVC
```

```python
model = SVC()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifer = GridSearchCV(model,{
    "C":[1,10,20,30],
    'kernel':['rbf','linear'],
},cv=5,return_train_score=False)
```

```python
classifer.fit(X,y)
```

```python
results= pd.DataFrame(classifer.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
classifer_r = RandomizedSearchCV(model,{
    "C":[1,10,20,30],
    'kernel':['rbf','linear'],
},cv=5,return_train_score=False,n_iter=5)
```

```python
classifer_r.fit(X,y)
```

```python
results= pd.DataFrame(classifer_r.cv_results_)
```

```python
results[['param_C','param_kernel','mean_test_score']]
```
